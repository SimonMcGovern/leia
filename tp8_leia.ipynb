Bucle de entrenamiento

```python
import numpy as np
from michigrad.michigrad.engine import Value
from michigrad.michigrad.visualize import show_graph
from michigrad.michigrad.nn import Neuron, MLP, Layer

np.random.seed(42)
xor = MLP(2, [2,1]) # Perceptrón multi capa. 2 entradas, capa oculta de 2 neuronas, 1 salida
entradas = [[0,0], [1,0], [0,1], [1,1]]
salidas = [0,1,1,0]
steps = 1000
learning_rate = 0.01
```

```python
for _ in range(steps):
    # Forward pass
    yhats = [xor(x).relu() for x in entradas]
    # Función de perdida
    loss = sum([(salida - yhat)**2 for salida, yhat in zip(salidas, yhats)])/4
    # Gradiente en 0
    for p in xor.parameters():
        p.grad = 0.
    # backward pass
    loss.backward()
    # Actualizar
    for p in xor.parameters():
        p.data -= p.grad * learning_rate

loss
```
